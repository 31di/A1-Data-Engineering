### Task 3 - Filter records
I remove bad records.  
the base is: valid record must have 9 columns and the salary must be a number.

```python
valid_employees = employees.filter(lambda e: len(e) == 9 and e[4].isdigit())
```

```python
print("Valid records:")
print(valid_employees.collect())
```

### Task 4 - Average salary per department
I take department and salary ,add all salaries and count employees,  
Then I divide total salary by count to get the average.

```python
dept_salary = valid_employees.map(lambda e: (e[2], (int(e[4]), 1)))
dept_totals = dept_salary.reduceByKey(lambda a, b: (a[0] + b[0], a[1] + b[1]))
avg_salary = dept_totals.mapValues(lambda v: v[0] / v[1])
```

```python
print("Average salary per department:")
print(avg_salary.collect())
```

### Task 5 - Employee count per department
I take department and put 1 for each employee, add them with `reduceByKey`. so This gives me the number of employees in each department.

```python
emp_count = valid_employees.map(lambda e: (e[2], 1)) \
                           .reduceByKey(lambda a, b: a + b)
```
